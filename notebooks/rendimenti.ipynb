```python
import numpy as np

def prezzo_bond(cedola, rendimento, anni, nominale=100):
    """Prezzo di un bond a cedola annuale: valore attuale di cedole + rimborso."""
    t = np.arange(1, anni + 1)                     # anni di pagamento cedole
    flussi = np.full(anni, cedola * nominale)      # cedole annue
    flussi[-1] += nominale                         # rimborso del capitale a scadenza
    return np.sum(flussi / (1 + rendimento) ** t)  # attualizzazione

def duration_modificata(cedola, rendimento, anni, dy=1e-4):
    """Duration stimata numericamente: variazione % del prezzo per +1 punto di tasso."""
    p0 = prezzo_bond(cedola, rendimento, anni)
    p_su = prezzo_bond(cedola, rendimento + dy, anni)
    p_giu = prezzo_bond(cedola, rendimento - dy, anni)
    return (p_giu - p_su) / (2 * p0 * dy)
```
